```python
from pathlib import Path
import os, subprocess, sys
ROOT = Path(r"C:\Users\ASUS\Documents\Quantum computing\Qiskit\GidNET-Qubit-Reuse-Algorithm")
BENCH = Path(r"C:\Users\ASUS\Documents\Quantum computing\Qiskit\CaQR-main\CaQR-main\benchmarks")
PREFERRED = Path(r"C:\Users\ASUS\caqr-qiskit045-env\Scripts\python.exe")
PY = str(PREFERRED) if PREFERRED.exists() else sys.executable
print("Runner Python:",PY)
for name, path in {"script":ROOT/"experiment4A_physical_value.py", "mutations":ROOT/"experiment4a_inputs"/"adversarial_mutation_cases.csv", "audit":ROOT/"experiment4a_inputs"/"width_and_original_plan_audit.csv", "plans":ROOT/"experiment2_results"/"experiment2_plan_features.csv", "benchmarks":BENCH}.items():
    print(name, path.exists(), path)
    assert path.exists(), f"Missing required {name}: {path}"
version_check = subprocess.run([PY,"-c","import qiskit; print(qiskit.__version__)"],cwd=ROOT,capture_output=True,text=True)
if version_check.returncode: raise RuntimeError("Qiskit is not installed in the selected Python: " + version_check.stderr)
print("Qiskit:",version_check.stdout.strip())
```

```python
def run_stage(stage, smoke_routes=0):
    cmd=[PY, str(ROOT/"experiment4A_physical_value.py"),stage]
    if stage!="prepare": cmd += ["--bench-dir",str(BENCH)]
    if smoke_routes: cmd += ["--smoke-routes",str(smoke_routes)]
    env=os.environ.copy()
    env["PYTHONPATH"]=str(ROOT)+os.pathsep+env.get("PYTHONPATH","")
    result=subprocess.run(cmd,cwd=ROOT,env=env)
    print("Return code:",result.returncode)
    if result.returncode: raise RuntimeError(f"{stage} failed")
```

## 1 — Prepare
این مرحله Routing ندارد. تعدادها باید مطابق Full Coverage باشند: **50 / 8642 / 485** و A/B/C برابر **130 / 289 / 291**.

```python
run_stage("prepare")
```

## 2 — Smoke test
فقط یک Route انجام می‌شود. اگر Import / Conversion خطا داد، سراغ Explore کامل نرو.

```python
run_stage("explore",smoke_routes=1)
```

## 3 — Full explore
این مرحله می‌تواند محاسباتی سنگین باشد. خروجی پس از هر Route ذخیره می‌شود و می‌توان همان فرمان را دوباره اجرا کرد.

```python
run_stage("explore")
```

## 4 — Validate
برندگان 4A را با seedهای مستقل (100–129) ارزیابی می‌کند. انتخاب مجدد انجام نمی‌شود.

```python
run_stage("validate")
```
